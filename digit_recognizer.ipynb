```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.decomposition import PCA
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
from tensorflow.keras import layers, models
```

```python
train = pd.read_csv('data/train.csv')
X_test = pd.read_csv('data/test.csv')
```

```python
X = train.drop('label', axis=1)
y = train['label']
```

```python
X = X / 255.0
X_test = X_test / 255.0
```

```python
X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.1, random_state=42, stratify=y)
```

```python
pca = PCA(n_components=50, random_state=42)
X_train_pca = pca.fit_transform(X_train)
X_val_pca = pca.transform(X_val)
X_test_pca = pca.transform(X_test)
```

```python
clf = SVC(kernel='rbf', C=10, gamma='scale')
clf.fit(X_train_pca, y_train)
pred = clf.predict(X_val_pca)
print('accuracy:', accuracy_score(y_val, pred))
```

```python
mlp = models.Sequential([
    layers.Input(shape=(784,)),
    layers.Dense(128, activation='relu'),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])
```

```python
mlp.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
mlp.fit(X_train, y_train, validation_data=(X_val, y_val), epochs=10, batch_size=64)
```

```python
loss, acc = mlp.evaluate(X_val, y_val)
print('MLP val accuracy:', acc)
```

```python
X_train_cnn = X_train.values.reshape(-1, 28, 28, 1)
X_val_cnn = X_val.values.reshape(-1, 28, 28, 1)
X_test_cnn = X_test.values.reshape(-1, 28, 28, 1)

cnn = models.Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Conv2D(32, 3, activation='relu'),
    layers.MaxPooling2D(),
    layers.Conv2D(64, 3, activation='relu'),
    layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

cnn.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
cnn.fit(X_train_cnn, y_train, validation_data=(X_val_cnn, y_val), epochs=10, batch_size=64)
```

```python
loss, acc = cnn.evaluate(X_val_cnn, y_val)
print('CNN val accuracy:', acc)
```

```python
pred_probs = cnn.predict(X_test_cnn)
preds = np.argmax(pred_probs, axis=1)
submission = pd.read_csv('data/sample_submission.csv')
submission['Label'] = preds
submission.to_csv('submission.csv', index=False)
```
